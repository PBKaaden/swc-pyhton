# analyze multiple files 

```python
import glob 

#using glob to find all infla files 

file_pattern = "data/inflammation-*.csv"
all_files = sorted(glob.glob(file_pattern))

# print(all_files)

data_arrays = []

for file in all_files:
    data = np.loadtxt(file, delimiter=",")
    data_arrays.append(data)
    print("wizzard file", file)

```

```python

fig = plt.Figure(figsize=[10,3])


for datasets in data_arrays:
    #mean, max, min per set 
    mean_pd = np.mean(datasets, axis=0)
    max_pd = np.max(datasets, axis=0)
    min_pd = np.min(datasets, axis=0)


    #make-y the plots   

    fig = plt.figure(figsize=[10,3])

    axis1 = fig.add_subplot(1,3,1)
    axis2 = fig.add_subplot(1,3,2)
    axis3 = fig.add_subplot(1,3,3)

    axis1.plot(mean_pd)
    axis2.plot(max_pd)
    axis3.plot(min_pd)

    axis1.set_ylabel("Mean value")
    axis2.set_ylabel("Max value")
    axis3.set_ylabel("Min value")

    for ax in [axis1, axis2, axis3]:
        ax.set_xlabel("Day")











```

# modular code test


### so you want to make function that are self understandable,e.g. use names that understand 
also make the function uni functional, bad practice to combine a bunch of stuff 

```python
def reading_data(filename, seperator_style = ","):
    """Reads numerical data from a text file with standard seperator of , for csv"""
    #cool did not now that that was the unit you created from this that shows up for the help function, coolio!
    return np.loadtxt(filename, delimiter=seperator_style)

def analyze_data(data, axis_input = 0):
    """ Takes numerical data in the form of a np array and calculates 
        mean, maximum, minimum, and standard deviation by row or column 
        with column as standard
        
    
        Parameters
        ----------
        data : numpy array
            numerical array data.
        axis_input : int 
            intereger input for the selection of column, 0, or row, 1, of the array.
    
        Returns  
        -------
        list
            list of mean, max, min, and std of row or column of numerical array.
    """
    return np.mean(data,axis=axis_input), np.max(data,axis=axis_input), np.min(data,axis=axis_input), np.std(data,axis=axis_input)

def visualize_data(analyzed_data):
    """   
        Optional longer description explaining the function.
    
        Parameters
        ----------
        parameter1 : type
            Description of the first parameter.
        parameter2 : type
            Description of the second parameter.
    
        Returns
        -------
        type
            Description of the returned value.
    
    """
    fig = plt.figure(figsize=[10,3])

    axis1 = fig.add_subplot(1,3,1)
    axis2 = fig.add_subplot(1,3,2)
    axis3 = fig.add_subplot(1,3,3)

    axis1.plot(analyzed_data[0])
    axis2.plot(analyzed_data[1])
    axis3.plot(analyzed_data[2]) 

    axis1.set_ylabel("Mean value")
    axis2.set_ylabel("Max value")
    axis3.set_ylabel("Min value")

    
    for ax in [axis1, axis2, axis3]:
        ax.set_xlabel("Day")

    return fig
```

```python
for file in all_files:
    data = reading_data(file)
    meanval,maxval,minval,stdval = analyze_data(data)
    fig = visualize_data([meanval,maxval,minval])
    fig.suptitle(file[5:])
```
