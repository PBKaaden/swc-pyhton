## Exercise: combining strings

"Adding" two strings produces their concatenation: `'a' + 'b'` is `'ab'`. Write a function called `fence` that takes two parameters called `original` and `wrapper` and returns a new string that has the wrapper character at the beginning and end of the original. A call to your function should look like this, and print `*name*`

`````python
def fahrenheit_to_celc(farh_input, rounding = 2, flip = False):
    if flip == False:
        return round(((farh_input - 32) * (5/9)), rounding)
    else:
        return round(((farh_input * (5/9))+32), rounding)
    
`````

`````python
farh_list = [ 99,100,101,40,80,50]

cels_list = []

for i in farh_list:
    cels = fahrenheit_to_celc(i)
    cels_list.append(cels)

print(cels_list)
`````

`````python


import numpy as np
import matplotlib.pyplot as plt


def analyze_data(data):

    maxval, minval, stdval,meanval = np.amax(data), np.amin(data), np.std(data),np.mean(data)
    return maxval, minval, stdval,meanval

def visualize_data(data):

    fig=plt.figure(figsize=(15.0, 3.0))
    axes1=fig.add_subplot(1, 3, 1)
    axes2=fig.add_subplot(1, 3, 2)
    axes3=fig.add_subplot(1, 3, 3)

    axes1.plot(np.mean(data, axis=0))
    axes2.plot(np.amax(data, axis=0))
    axes3.plot(np.amin(data, axis=0))

    for ax in [axes1,axes2,axes3]:
        ax.set_xlabel('day')

    axes1.set_ylabel('mean')
    axes2.set_ylabel('max')
    axes3.set_ylabel('min')

    return fig
`````
